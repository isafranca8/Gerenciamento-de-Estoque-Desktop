```python
#NEGÓCIO CERTO
import tkinter as tk
janela=tk.Tk()
janela.title("Tela de loguin")
janela.geometry("500x350")
janela.iconbitmap("logoo-ofcc.ico")

#frame superior com pack
frame_topo= tk.Frame(janela, bg="#236591", height=50)
frame_topo.pack(fill="x", pady=10)
tk.Label(frame_topo, text="Estoque pro", bg="#236591", fg="white", font=("Arial", 16)).pack(pady=10)
tk.Label(frame_topo, text="Controle total, resultado real", bg="#236591", fg="white", font=("Arial", 12)).pack(pady=10)
img_tk = tk.PhotoImage(file="logoo ofcc.png")
img_tk = img_tk.subsample(3)

#Frame de conteúdo principal com grid
frame_conteudo= tk.Frame(janela, bg="#C2C2C2")
frame_conteudo.pack(fill="both", expand=True, padx=5, pady=5)

#configurações de grid para o frame de conteúdo
frame_conteudo.grid_rowconfigure(0, weight=9)
frame_conteudo.grid_rowconfigure(1, weight=2)
frame_conteudo.grid_columnconfigure(0, weight=9)
frame_conteudo.grid_columnconfigure(1, weight=1)

#Fame cinza para hospedar loguin
frame_azul_conteudo = tk.Frame(frame_conteudo, bg="#C2C2C2")
frame_azul_conteudo.grid(row=0, columnspan=2, padx=5, pady=5, sticky="nsew")

frame_azul_conteudo.grid_columnconfigure(0, weight=1) # Topo
frame_azul_conteudo.grid_columnconfigure(2, weight=1) # Embaixo
frame_azul_conteudo.grid_rowconfigure(0, weight=1) # Esquerda
frame_azul_conteudo.grid_rowconfigure(2, weight=1) # Direita

frame_azul_conteudo.grid_columnconfigure(1, weight=1)
frame_azul_conteudo.grid_rowconfigure(1, weight=1)

#frame logiun
frame_form = tk.Frame(frame_azul_conteudo, bg="#236591")
frame_form.grid(row=1, column=1, sticky="n")
frame_form.grid_columnconfigure(2, weight=1)
frame_form.grid_rowconfigure(1, weight=1)
frame_form.grid_rowconfigure(2, weight=1)
frame_form.grid_rowconfigure(3, weight=1)
frame_form.grid_rowconfigure(4, weight=1)

#Objetos
tk.Label(frame_form, text="Usuário:", bg="#236591", fg="white", font=("Arial", 12)).grid(row=1, column=1, padx=10, pady=5, sticky="w")
entrada_usuario = tk.Entry(frame_form, width=30)
entrada_usuario.grid(row=1, column=2, padx=10, pady=5, sticky="nsew")

# Senha
tk.Label(frame_form, text="Senha:", bg="#236591", fg="white", font=("Arial", 12)).grid(row=2, column=1, padx=10, pady=5, sticky="w")
entrada_senha= tk.Entry(frame_form, width=30, show="*")
entrada_senha.grid(row=2, column=2, padx=10, pady=5, sticky="nsew")

#Escolha de funcionário ou administrador
cargo_selecionado = tk.StringVar(frame_form, "Funcionário")
tk.Label(frame_form, text="Cargo:", bg="#236591", fg="white", font=("Arial", 10)).grid(row=3, column=1, padx=10, pady=5, sticky="w")

frame_radio = tk.Frame(frame_form, bg="#236591")
frame_radio.grid(row=3, column=2, padx=10, pady=5, sticky="w")

tk.Radiobutton(frame_radio,
               text="Funcionário",
               variable=cargo_selecionado,
               value="Funcionário",        # Define o valor desta opção
               bg="#236591", fg="white", font=("Arial", 10)).grid(row=3, column=3, padx=(10, 0), pady=5, sticky="w")

tk.Radiobutton(frame_radio,
               text="Administrador",
               variable=cargo_selecionado,
               value="Administrador",      # Define o valor desta opção
               bg="#236591", fg="white", font=("Arial", 10)).grid(row=3, column=4, padx=(100, 0), pady=5, sticky="w")

# Botão Login
tk.Button(frame_form, text="Login", font=("Arial", 12)).grid(row=4, column=1, columnspan=2, pady=10, sticky="n")




janela.mainloop()

```

```python
import tkinter as tk
from tkinter import messagebox

janela = tk.Tk()
janela.title("Seleção Múltipla")

opcoes = ["Opção A", "Opção B", "Opção C", "Opção D", "Opção E"]

def mostrar_selecao():
    selecionados = []
    # Obtém os índices de todos os itens selecionados
    indices_selecionados = listbox.curselection()
    
    for i in indices_selecionados:
        # Obtém o valor real do item pelo seu índice
        selecionados.append(listbox.get(i))
        
    messagebox.showinfo("Seleção", f"Itens Selecionados: {', '.join(selecionados)}")

## Criação da Listbox
# 'selectmode=tk.MULTIPLE' é a chave para a seleção múltipla
listbox = tk.Listbox(
    janela, 
    selectmode=tk.MULTIPLE,
    height=len(opcoes), # Define a altura para mostrar todas as opções
    font=("Arial", 12)
)
listbox.pack(pady=10, padx=20)

# Inserindo as opções
for item in opcoes:
    listbox.insert(tk.END, item)

# Botão para mostrar o resultado da seleção
tk.Button(janela, text="Ver Seleção", command=mostrar_selecao).pack(pady=10)

janela.mainloop()
```
